```python
import matplotlib.pyplot as plt
import seaborn as sns

db_tips = sns.load_dataset('tips')
print(db_tips)

sns.scatterplot(data=db_tips, x='total_bill', y='tip', hue='smoker')

plt.title('Gorjeta x Conta')
plt.xlabel('Total da conta')
plt.ylabel('Gorjeta')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

db_penguins = sns.load_dataset('penguins')

print(db_penguins)

sns.histplot(data=db_penguins, x='flipper_length_mm', hue='species', kde=True)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

db_tips = sns.load_dataset('tips')

#print(db_tips)

sns.boxplot(data=db_tips, x='day', y='tip', hue='sex')

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

db_tips = sns.load_dataset('tips')

print(db_tips)

sns.regplot(data=db_tips, x='total_bill', y='tip', line_kws={'color':'green'})

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

db_tips = sns.load_dataset('tips')

# calculando a matriz de correlação
correlacao = db_tips[['total_bill', 'tip', 'size']].corr()
print(correlacao)

sns.heatmap(data=correlacao, annot=True, fmt='.2f')

plt.show()
```
