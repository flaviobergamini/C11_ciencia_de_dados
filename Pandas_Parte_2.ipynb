** Calculando qual a porcentagem das populações dos países representam na população mundial**

```python
import numpy as np

total_population = np.sum(dfpaises['Population'])

#print(total_population)

serie_percent_paises = (dfpaises['Population']/total_population) * 100
#print(np.round(serie_percent_paises, 3))

print(dfpaises.columns)

dfpaises['% Population'] = np.round(serie_percent_paises, 3)

print('----------------------')
print(dfpaises.columns)

dfpaises.to_csv('paises2.csv', sep=';')
```

**Excluindo coluna**

```python
print(dfpaises.columns)

print('--------------------------')

# A operação de exclusão é de fato aplicada no dataframe atual (dfpaises),
# se o parâmetro inplace=True estiver presente
dfpaises.drop('% Population', axis=1, inplace=True) # Se não encontrar a coluna, gera um erro

print(dfpaises.columns)
```

**Quero mostrar quantos paises tem por região, usando o agrupamento por região**

```python
group_region = dfpaises.groupby('Region')
print(group_region.count()['Country'])
```

**Somando a população agrupada pela Região**

```python
group_region = dfpaises.groupby('Region')
print(group_region.sum()['Population'])
```

**usando o método describe() dos agrupamentos para extrair métricas**

```python
group_region= dfpaises.groupby('Region')
print(group_region.describe())
```

**Criando funções próprias e usando no pandas**

```python
def percent(x):
  return x*0.9


deathrate1 = dfpaises['Deathrate']
print(deathrate1) # antes

print('--------------------')

# aplicando função customiada no pandas
deathrate2 = dfpaises['Deathrate'].apply(percent)

print(deathrate2) # depois
```
