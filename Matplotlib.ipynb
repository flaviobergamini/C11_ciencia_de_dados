```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4])
y = x*2
y2 = x*x

plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Gráfico de Linha')

plt.plot(x, y, 'r-.', x, y2, 'b:') # duas retas no mesmo plano cartesiano

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4])
y = x*2
y2 = x*x

plt.subplot(1, 2, 1) #qtd total de linhas, qtd total de colunas, posição 1 para o subgráfico
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Linha')
plt.plot(x, y, 'r-.')

plt.subplot(1, 2, 2) #qtd total de linhas, qtd total de colunas, posição 2 para o subgráfico
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Exponecial')
plt.plot(x, y2, 'b:')

plt.show() # mostra os dois gráficos em uma linha e em duas colunas
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4])
y = x*2
y2 = x*x

plt.subplot(2, 2, 1) #qtd total de linhas, qtd total de colunas, posição 1 para o subgráfico
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Linha')
plt.plot(x, y, 'r-.')

plt.subplot(2, 2, 2) #qtd total de linhas, qtd total de colunas, posição 2 para o subgráfico
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Exponecial')
plt.plot(x, y2, 'b:')

plt.subplot(2, 2, 3) #qtd total de linhas, qtd total de colunas, posição 1 para o subgráfico
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Linha')
plt.plot(x, y, 'y--')

plt.subplot(2, 2, 4) #qtd total de linhas, qtd total de colunas, posição 2 para o subgráfico
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Exponecial')
plt.plot(x, y2, 'k-')

# Calcula automaticamente os espaços necessários entre os elementos dos gráficos
# evitando sobreposição de informações
plt.tight_layout()

plt.show() # mostra os quatro gráficos em duas linhas e em duas colunas
```

```python
import matplotlib.pyplot as plt
import pandas as pd

dfpaises = pd.read_csv('paises.csv', delimiter=';')

# extraindo os 6 paises com maior extensão territorial
paises = dfpaises.nlargest(6, 'Area (sq. mi.)')

# definindo gráfico de dispersão com tamanho de circulos baseados no tamanho na extensão territorial
plt.scatter(paises['Country'], paises['Area (sq. mi.)'], s=paises['Area (sq. mi.)']/10000)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

dfpaises = pd.read_csv('paises.csv', delimiter=';')

# extraindo os 6 paises com maior extensão territorial
paises = dfpaises.nlargest(6, 'Area (sq. mi.)')

tamanho = paises['Area (sq. mi.)']/10000

# definindo gráfico de dispersão com tamanho de circulos baseados no tamanho na extensão territorial
# comparando a PIB per capita
plt.scatter(paises['Country'], paises['GDP ($ per capita)'], s=tamanho)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

dfpaises = pd.read_csv('paises.csv', delimiter=';')

# extraindo os 5 paises com maior PIB per capita
paises = dfpaises.nlargest(5, 'GDP ($ per capita)')

# um dataframe do pandas é composto por séries, vamos extrair duas séries
country = paises['Country']
gdp = paises['GDP ($ per capita)']

# desenhando um gráfico de barras
plt.bar(country, gdp, color='blue')

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

dfpaises = pd.read_csv('paises.csv', delimiter=';')

# máscara para filtrar paises que não possuem costa marítima
filter = dfpaises['Coastline (coast/area ratio)'] == 0

print(dfpaises.size) # a quantidade total de células no dataframe
print(len(dfpaises)) # a quantidade de linhas do dataframe
print(len(dfpaises.columns) * len(dfpaises)) # calcula a quantidade total de células no dataframe

# extraindo os paises que não possuem costa marítima
paises = dfpaises[filter]

qtNoCoast = len(paises)
qtdCoast = len(dfpaises) - qtNoCoast

# desenhando o gráfico de pizza
plt.pie(x=[qtdCoast, qtNoCoast], labels=['% Países com Costa Marítima', '% Países sem Costa Marítima'], autopct='%1.1f%%')

# Para exibir as porcentagens, é adicionado o parâmetro autopct='%1.1f%%'
# O valor %1.1f%% define a formatação
# - A primeira % define que é um valor numérico
# - O primeiro valor 1, define a largura mínima, "como se fosse um vetor com tamanho ajustável"
# - O terceiro valor que é um conjunto .1f define que é um ponto flutuante (float) com uma casa de precisão
# - O último valor %% exibe o caractere de %

plt.show()
```
