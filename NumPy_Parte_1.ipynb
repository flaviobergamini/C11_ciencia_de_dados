**Operações básicas**

```python
arr = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])

print(f'Valor mínimo: {arr.min()}, índice: {arr.argmin()}')
print(f'Valor máximo: {arr.max()}, índice: {arr.argmax()}')
```

```python
# Exemplo Extra com trigonometria, mostrando a superioridade do numpy trabalhando com arrays e não somente com scalar necessitando de um loop tradicional e pouco performático
x = 1
y = 5

distancia_cm =np.hypot(x, y) * 100
angulo_graus = np.degrees(np.arctan2(y,x))

print(distancia_cm)
print(angulo_graus)

print()
import math

dist = math.sqrt(x**2 + y**2)*100
angulo = math.atan2(y,x)*(180/math.pi)
print(dist)
print(angulo)
```

```python
print(f'Soma: {arr.sum()}, média: {arr.mean()}')
print(arr)
arr2 = np.array([5,6,7,8])
print(arr2)

print(f"Soma índice a índice: {arr + arr2}")
print(f'Multiplicação índice a índice: {arr * arr2}')
print(f' Juntar dois arrays: {np.concatenate((arr, arr2))}')
```

**Remodelando forma do array**

```python
arr = np.arange(9)
print(arr)

print()

matriz = arr.reshape([3,3])  # Precisa ter o mesmo tamanho do array unidimensional
print(matriz)

print()
print(f'Tamanho array: {matriz.size}')
print(f'Quantidade de dimensões: {matriz.ndim}')
print(f'Qual o corpo, modelo do array? {matriz.shape}')
print(f'Soma das colunas: {matriz.sum(axis=0)}')
print(f'Soma das linhas: {matriz.sum(axis=1)}')
print('Operações entre matriz e um valor escalar (Broadcast):')
print(matriz*5)
print()
print(matriz/2)
print()
print(matriz-5)
```

**Números aleatórios**

```python
np.random.seed(5)

arr = np.random.randint(0, 10, 10)
print(arr)

matriz = np.random.randint(0, 10, [5, 5])
print(matriz)
```

**Identifica únicos e conta recoreências de repetições**

```python
arr = np.array([1,2, 2, 3, 3, 4])

print(np.unique(arr))

print(np.unique(arr, return_counts=True))

```
