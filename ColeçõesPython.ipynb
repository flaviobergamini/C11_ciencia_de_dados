**Listas**

```python
nomes = ['Goku', 'Vegeta', 'Trunks', 'Gohan']
print(nomes)
```

Inserindo dados na lista

```python
nomes.append('Gohan') # inserindo no final da lista
```

```python
nomes.insert(2, "Kuririn") # Insere em uma posição específica
```

```python
print(nomes)
```

Removendo itens da lista

```python
del nomes[4]
```

```python
nomes.pop(3)
```

```python
if 'Gohan' in nomes:
  nomes.remove('Gohan')
  print('Removido com sucesso!!')
else:
  print('Não está na lista')
```

```python
tamanho_lista = len(nomes)

print(tamanho_lista)

if tamanho_lista > 4:
  nomes.pop(4)
  print('item removido pelo índice!')
else:
  print('Lista muito pequena, não removido')
```

Ordenação

**Tuplas**

```python
nomes_tupla = ('Goku', 'Vegeta', 'Trunks', 'Gohan')
print(nomes_tupla)
print(type(nomes_tupla))
print(type(nomes))
```

```python
print(nomes_tupla[0])
print(nomes_tupla[1:3])
```

```python
print(nomes_tupla)
nomes_tupla[2] = "Bulma"
```

```python
for nome in nomes_tupla:
  print(nome)

print('----------------')

for count in range(len(nomes_tupla)):
  print(f'Índice: {count}, Valor: {nomes_tupla[count]}')
```

Operações com tuplas

```python
a = (2, 6, 8)
b = (4, 6, 9, 5, 6)
z = a + b

print(z)

print(min(z), max(z))

print(z.count(6)) # quantidade de repetições de um mesmo valor em uma tupla

print(z.index(9)) # o índice de um valor na tupla
```

**Conjunto (Set)**

```python
nomes_set = {'Goku', 'Vegeta', 'Trunks', 'Gohan', 'Gohan', 'Gohan', 'Trunks'}
print(nomes_set)
print(type(nomes_set))
nomes_set.remove('Trunks')
nomes_set.add('TRUNKS')
print(nomes_set)
```

Operações com os conjuntos

```python
a = {2, 6, 8}
b = {1, 6, 5}

# União
z = a | b
print(f'União: {z}')

# Diferença (mostra a diferença do primeiro em relação a segundo)
z = a - b
z1 = b - a
print(f'Diferença z: {z}')
print(f'Diferença z1: {z1}')

#Interseção
z = a & b
print(f'Interseção: {z}')
```
