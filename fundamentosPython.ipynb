**Concatenações**

```python
print(True, False)
print("PI=", 3.1415)
print("PI=" + "3.1415")
print("PI={}, Euler={}".format(3.1415, 2.71))
print("PI={0}, Euler={1}".format(3.1415, 2.71))
print("PI={1}, Euler={0}".format(3.1415, 2.71))
print(f'PI={3.14157788888}')

pi = f'\nPI={3.14157788888}'
num = "PI={1}, Euler={0}".format(3.1415, 2.71)

print(pi)
print(num)
```

**Tipos de Dados**

```python
a = 3
b = 2.71
c = True
d = 'Inatel'

print(type(a))
print(type(b))
print(type(c))
print(type(d))

# Casting
e = '3'
print('----------')
print(type(e))
e = int(e)
print(type(e))
```

**Entradas de dados com o teclado**

```python
nome = input('Entre com o seu nome: ')
ano_nasc = int(input('Entre com seu ano de nascimento: '))
peso = float(input('Entre com o seu peso: '))

print('Nome, tipo: ', type(nome), 'Conteúdo: ', nome)
print('Ano, tipo:', type(ano_nasc), 'Conteúdo:', ano_nasc)
print('Peso, tipo:', type(peso), "Conteúdo: ", peso)
```

**Operadores Aritméticos**

```python
a = 10
b = 3

print('a=', a)
print('b=', b)

# Operadores aritméticos
print('\nSoma:', a + b)
print('Subtração:', a - b)
print('Multiplicação:', a * b)
print('Divisão:', a / b)
print('Potenciação:', a ** b)
print('Divisão inteira:', a // b)
print('Resto da divisão:', a % b)

# Operadores de comparação
print("a == b", a == b)
print("a == 10", a == 10)

#Precedência dos operadores
print("2 + 3 * 4 = ", 2 + 3 * 4)
print('(2 + 3) * 4 = ', (2 + 3) * 4)
```

**Importação de Bibliotecas**

```python
import math

num = 2.3
print(math.ceil(num))
print(math.floor(num))
print(math.trunc(num))
```

**Manipulando cadeia de caracteres**

```python
palavra = 'Hello World'

print('String:', palavra)

print(palavra[6])
print(palavra[:5])
print(palavra[6:11])
print(palavra[3:])
print(palavra[0:11:2])
print(palavra[::-1])

print('World' in palavra)

palavra = palavra.replace('World', 'Python')
print(palavra)
```

**Estrutura de decisão**

```python
x = int(input("Entre com  um número:"))

if x < 0:
  print('Negativo')
elif x == 0:
  print('Igual a zero')
else:
  print('Positivo')
```

**Estrutura de Repetição**

```python
for i in range(10):
  print(i)

print('--------------')

for i in range(-5, 10):
  print(i)

print('--------------')

for i in range(10, 0, -1):
  print(i)

print('--------------')

for i in range(10, 0, -2):
  print(i)

print('--------------')

coisas = ['Cachorro', True, 3, 2.71]
for i in coisas:
  print(i)
```
